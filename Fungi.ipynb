## My Question:
## What is the most frequently observed odor of poisnes mushrooms?

## My Study:
## I will compare divide all the smells into differnt bins, and then figure out the precentage of that smell leading to the fungi being poisness.

```python
import pandas as pd
```

```python
mush = pd.read_csv("mushrooms.csv")
```

```python
odor_labels = {'a': 'Almond', 'l': 'Anise',  'c': 'Creosote', 'y': 'Fishy', 'f': 'Foul', 'm': 'Musty', 'n': 'None', 'p': 'Pungent', 's': 'Spicy'}
mush['odor_labels'] = mush['odor'].map(odor_labels)
```

```python
poisonous = mush[mush['class'] == 'p']
```

```python
odor_counts = poisonous['odor_labels'].value_counts()
print(odor_counts)
```

```python
import seaborn as sns
```

```python
sns.barplot(x=odor_counts.index, y=odor_counts.values, order=odor_counts.index)
```

## Conclusion

## The most frequently observed odor of poisness mushrooms is foul. I know this because the foul smell has the highest mode of 2160 mushrooms found that smell foul and turned out to be poisonous.
