## Question: 
Do Colorado School of Mines students earn more after 10 years compared to South Dakota School of Mines students? And how does there Z scores compare to those of other postsecondary institutions in the United States?

## Type of study 
My study is a observational study and census. It is both because my first question I am asking only involves 2 schools and 2 data tables which makes its observational, but my second question I am asking is the whole date set which would make a census.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("college_data.csv")
df.info()
```

```python
mean = df["median_earnings_10yrs"].mean()
median = df["median_earnings_10yrs"].median()
std = df["median_earnings_10yrs"].std()
print(mean, median)
```

```python
csm = df[df["school_name"] == "Colorado School of Mines"]
sdsm = df[df["school_name"] == "South Dakota School of Mines and Technology"]

csm_earnings = csm["median_earnings_10yrs"].values[0]
sdsm_earnings = sdsm["median_earnings_10yrs"].values[0]

print(csm_earnings, sdsm_earnings)
```

```python
mean_earnings = df["median_earnings_10yrs"].mean()
std_earnings = df["median_earnings_10yrs"].std()
```

```python
print(mean_earnings, std_earnings)
```

```python
csm_z = (csm_earnings - mean_earnings)/ std_earnings
```

```python
sdsm_z = (sdsm_earnings - mean_earnings)/ std_earnings
```

```python
print(csm_z, sdsm_z)
```
