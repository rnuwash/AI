## Question:
What release year has the highest avg number of streams?

## This is an observational study because I am only using a sample from 2023's most popular songs on spotify. I know this because I am only using a sample data and am not using everysingle song on spotify.

```python
import pandas as pd
```

```python
spotify = pd.read_csv("spotify.csv", encoding="latin-1")
```

```python
spotify.head()
```

```python
spotify["streams"] = pd.to_numeric(spotify["streams"], errors="coerce")
spotify.groupby("released_year")["streams"].mean()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
avg_streams = spotify.groupby("released_year")["streams"].mean().reset_index()
avg_streams.columns = ["released_year", "streams"]
plt.figure(figsize=(16, 6))
sns.barplot(data=avg_streams, x="released_year", y="streams")
plt.axhline(spotify["streams"].mean(), color="red", linestyle="--", label="Overall Mean")
plt.axhline(spotify["streams"].median(), color="pink", linestyle="--", label="Overall Median")
plt.xticks(rotation=90)
plt.tight_layout()
plt.legend()
plt.show()
```

```python
avg_streams_sorted = avg_streams.sort_values("streams", ascending=False)

plt.figure(figsize=(16, 6))
sns.barplot(data=avg_streams_sorted, x="released_year", y="streams",
            order=avg_streams_sorted["released_year"])
plt.axhline(spotify["streams"].mean(), color="red", linestyle="--", label="Overall Mean")
plt.axhline(spotify["streams"].median(), color="pink", linestyle="--", label="Overall Median")
plt.xticks(rotation=90)
plt.tight_layout()
plt.legend()
plt.show()
```
