```python
import pandas as pd

df = pd.read_csv('titanic.csv')

# Bin ages into groups
bins = [0, 12, 18, 30, 45, 60, 100]
labels = ['Child (0-12)', 'Teen (13-18)', 'Young Adult (19-30)', 
          'Adult (31-45)', 'Middle Age (46-60)', 'Senior (60+)']
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)

# Survival rate and count per age group
survival_by_age = df.groupby('AgeGroup', observed=True)['Survived'].agg(['mean', 'count'])
print(survival_by_age)

# Simple linear correlation (weak on its own, since effect isn't linear)
print(df['Age'].corr(df['Survived']))
```

# My Conclusion
## The age did affect the survival rate. If you were in the age's between 19-45 you had a much higher chance of survuival.
