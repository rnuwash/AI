```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
rng = np.random.default_rng()  
```

```python
sample_space = ["One", "Two", "Three", "Four", "Five", "Six"]

print("Sample space for one coin flip:", sample_space)
print("Number of possible outcomes:", len(sample_space))
```

```python
outcomes_in_event = 1
total_outcomes = len(sample_space)

theoretical_one = outcomes_in_event / total_outcomes
print("Theoretical P(one) =", theoretical_one)
```

```python
def roll_dice(number_of_rolls):
    values = rng.integers(1, 7, size=number_of_rolls)


    # Put the results in a dataframe, numbering each flip
    return pd.DataFrame({
        "Roll_number": np.arange(1, number_of_rolls + 1),
        "value": values,
    })

# Flip 10 coins and look at every result
rolls_1000 = roll_dice(1000)
rolls_1000
```

```python
counts_1000 = rolls_1000["value"].value_counts()

print("Counts of each outcome:")
print(counts_1000)
print()
print("Total number of flips:", counts_1000.sum())
```

```python
one_counts = (rolls_1000["value"] == 1).sum()
total = len(rolls_1000)

relative_freq_heads = one_counts / total

print(f"One came up {one_counts} times out of {total} rolls.")
print(f"Relative frequency of One = {one_counts}/{total} = {relative_freq_heads}")
print(f"Theoretical probability of One = {theoretical_one}")
```

```python
# Set visual style
sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))
ax = sns.countplot(data=rolls_1000, x="value", color="#3498db")

plt.axhline(
    y=1000 / 6,
    color="red",
    linestyle="--",
    linewidth=1.5,
    label="Theoretical Expected Count (~166.7)",
)

plt.title(
    "Frequency of Each Outcome in 1,000 Dice Rolls",
    fontsize=14,
    pad=15,
)
plt.xlabel("Die Outcome", fontsize=12)
plt.ylabel("Count / Frequency", fontsize=12)
plt.legend()

for p in ax.patches:
    ax.annotate(
        f"{int(p.get_height())}",
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="bottom",
        xytext=(0, 3),
        textcoords="offset points",
    )

plt.tight_layout()
plt.show()
```
