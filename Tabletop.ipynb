## My Question
***Write your question here***. On a 5x5 board with 3 hidden ships, how many turns does it take on average to sink all 3 ships by guessing randomly without repeating?

## My Simulation
Describe your simulation here:
1. Random Variable ($X$): Total turns needed to find all 3 ships ($X$ can be any whole number from 3 to 25).
2. Trial and Outcomes: One trial is playing one full game. The outcome is the total number of turns taken.

### My Analysis

```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
results = []
```

```python
for i in range(1000):
    board = list(range(1, 26))  # 25 grid squares
    ships = random.sample(board, 3)  # 3 hidden ships
    guesses = random.sample(board, 25)  # Random guess order
    
    hits = 0
    turns = 0
    
    for guess in guesses:
        turns += 1
        if guess in ships:
            hits += 1
        if hits == 3:
            break  # Stop when all 3 ships are hit
            
    results.append(turns)
```

```python
mean_turns = np.mean(results)
std_turns = np.std(results)
```

```python
print(f"Mean Turns: {mean_turns:.2f}")
print(f"Standard Deviation: {std_turns:.2f}")
```

```python
plt.hist(results, bins=range(3, 27), density=True, edgecolor='black', color='teal')
plt.title("Turns to Win Battleship")
plt.xlabel("Turns (X)")
plt.ylabel("Probability")
plt.show()
```

## My Conclusion
1. The expected value (mean) is 19.59 turns, and the standard deviation is 4.38 turns.
2. Graph Interpretation: Winning in 3 turns is very rare ($<0.1\%$). Most games end between turns 18 and 23 because most of the board has been searched by then.
3. Support: The graph peaks near turn 19.5, showing you usually have to guess through most of the 25 squares to find all 3 ships
