## Question?
What is the **joint probability** that you have **all 4 aces in the tableau are showing** when setting up **flip 3 solitaire**? 

## My Simulation
    The random proccess that I am simulating is the set up of solitaire. You flip the first card in the first pile, then you move to the second pile and put one down face down and flip the 3 total card. i will simulate this for 7 collums adding one more card face down for every row.

    The trial will be testing to see if one of those cards flipped is an Ace, for all 7 collums, If all 4 aces are out then it is true. The possible outcomes could be any of the 52 cards. The compound event I will be studing is when an ace is flipped over, and if all 4 of them are shown. For example {Ace, King, Queen, Ace, Ace, Jack, Ace}, {King, Ace, Two, Ace, Ace, Jack, Ace}, {Three, Seven, Ace, Ace, Ace, Eight, Ace}
    
    The outcomes that make it true is when there is 4 aces in the spots 1, 3, 6, 10, 15, 21, 28. When 4 of these spots in the deck are occupied with an ace then it is true.

    I will call the event of getting an Ace in any of the 7 spots "1" and the event of it not being an ace I will call it "0".

    My question Is asking P(1 ^ 1 ^ 1 ^ 1) this is join probability.

    Events 1 and 0 are not mutally exclusive because I can still get any other card. 

    They are very dependant, Because they need all aces to show, its also lowers the chance or probabilty if the first card isnt an ace.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import random
```

```python
rng = np.random.default_rng() 
```

```python
exposed_indices = [0, 2, 5, 9, 14, 20, 27]
```

```python
deck_positions = list(range(52))
```

```python
num_simulations = 100_000
success_count = 0
```

```python
exposed_set = {0, 2, 5, 9, 14, 20, 27}
success_count = 0

for _ in range(num_simulations):
    deck_positions = rng.permutation(52)
    if set(deck_positions[:4]).issubset(exposed_set):
        success_count += 1
```

```python
simulated_prob = success_count / num_simulations
```

```python
print("Simulated Probability:", simulated_prob)
print("Percentage:", simulated_prob * 100, "%")
```

```python
theoretical_prob = (7 * 6 * 5 * 4) / (52 * 51 * 50 * 49)
```

```python
relative_error = abs(simulated_prob - theoretical_prob) / theoretical_prob * 100
```

```python
print("Simulated Probability:  ", simulated_prob)
print("Theoretical Probability:", theoretical_prob)
print("Relative Error:         ", relative_error, "%")
```

## Analysis

```python
sns.set_theme(style="whitegrid")
```

```python
labels = ["Simulated", "Theoretical"]
probabilities = [simulated_prob, theoretical_prob]
```

```python
plt.figure(figsize=(7, 5))
ax = sns.barplot(x=labels, y=probabilities,)
plt.title("Simulated vs Theoretical Probability", fontsize=14)
plt.ylabel("Probability", fontsize=12)
```

## In the graph above you can see the values I got when I simulated 100,000 games vs the theretical math that I have found. The error between the two is 8.29%.

After simulating 100_000 games the **joint probability** that you have **all 4 aces in the tableau are showing** when setting up **flip 3 solitaire** is .01399% chance that you will get all 4 aces face up in the tableau. Which is 8.28% off of the theretical amount.
